```python
## output formatting

1.using old style method formatting
2.using modulo operator
3.using f - Format method
4.using dot Format method
```

```python
## Using old style method formatting [using comma]

name = "phanindra"
age = 23
print("name is ",name ," and age is " ,(age))

```

```python
## Using modulo ooperator
## - %d - To represent integer 
## - %f - To represent flaot
## - %s - To represent string

name = "phanindra"
age = 23
# name is phanindra and age is 23
print("name is %s and age is %d"%(age , name))
```

```python
name = "phanindra"
age = 23
# name is phanindra and age is 23
print("name is %s and age is %d"%(name , age))
```

```python
name = "phanindra"
age = 23    ## converting int to str
# name is phanindra and age is 23
print("name is %s and age is %s"%(name , age))
```

```python
## adding list of numbers and searching for result
name = "phanindra"
age = [2,4,7,9]
# name is phanindra and age is 23
print("name is %s and age is %s"%(name , age))
```

```python
## Using Dot format method
name = "phanindra"
age = 23
# name is phanindra and age is 23
print("name is {} and age is {}".format(name , age))
```

```python
## Using f- format method
name = "phanindra"
age = 23
# name is phanindra and age is 23
print(f"name is {name} and age is {age}")
```

```python
## string too integer , float boolean , list conversions
val = str(input())
print(val)
print("flaot:",float(val))
print("integer:",int(val))
print("boolean:",bool(val))
print("list:",list(val))
print("tuple:",tuple(val))
print("set:",set(val))
```

```python
## list to string ,tuple and set conversions
list_nums =list(map(int,input().split()))

print(list_nums)

string_list =str(list_nums)

print(string_list)

print(string_list[:3])

print(string_list[:-1])

print(string_list[:1])

print("tuple:",tuple(list_nums))

print("set:",set(list_nums))
```

```python
### tuple to list, string, and set conversions
t = tuple(map(int,input().split()))
print(t)
num = list(t)
print(num)
print("string:",str(t))
print("set:",set(t))
```

```python
## set to string ,list and tuple
s = set(map(int,input().split()))
print("string conv:",str(s))
print("list conv:",list(s))
print("tuple conv:",tuple(s))
```
